```python
import pandas as pd
```

```python
df = pd.read_csv("data/g42_Exchange_Programs_v2.csv", sep = ';', parse_dates=['university_created_date','course_start_date'], dayfirst=True)
```

```python
df_university = df.loc[:,['university_id','university_name','university_created_date']]
```

```python
df_university = df_university.drop_duplicates(subset='university_id', keep='first')
```

```python
df_program = df.loc[:,['program_id','program_title','program_category']]
```

```python
df_program = df_program.drop_duplicates(subset='program_id', keep='first')
```

```python
df_program = df_program.dropna()
```

```python
df_faculty = df.loc[:,['faculty_id','faculty_info','university_id']]
```

```python
df_faculty = df_faculty.drop_duplicates(subset='faculty_id', keep='first')
```

```python
df_faculty = df_faculty.dropna()
```

```python
df_partnership = df.loc[:,['university_id','program_id','course_start_date', 'students_number']]
```

```python
df_partnership = df_partnership.drop_duplicates(subset=['university_id', 'program_id'], keep='first')
```

```python
df_partnership.insert(0, 'id', range(1, len(df_partnership) + 1))
```

```python
df_partnership = df_partnership.dropna()
```

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
creator = create_engine('sqlite:///data/g42_database.sqlite')
```

```python
df_university.to_sql('University', con=creator, if_exists='replace', index=False)
```

```python
df_program.to_sql('Program', con=creator, if_exists='replace', index=False)
```

```python
df_faculty.to_sql('Faculty', con=creator, if_exists='replace', index=False)
```

```python
df_partnership.to_sql('Partnership', con=creator, if_exists='replace', index=False)
```

```python
df_university.to_csv('university.csv', sep=';',index = False)
```

```python
df_program.to_csv('program.csv', sep=';',index = False)
```

```python
df_faculty.to_csv('faculty.csv', sep=';',index = False)
```

```python
df_partnership.to_csv('partnership.csv', sep=';',index = False)
```

```python
creator.dispose()
```
